# TicTacDoh — Negamax vs. Negamax with Alpha-Beta Pruning

This project implements **TicTacDoh**, a variant of Tic-Tac-Toe with an element of
randomness: in the *probabilistic* variant, each chosen move has a **20% chance of
failing**, in which case the player loses their turn.

Two search algorithms are compared across both variants:

1. **Negamax** — plain game-tree search
2. **Negamax with alpha-beta pruning** — same result, pruned search

For each algorithm, variant, and search depth, 1000 self-play games are run
(alternating the starting player) and win rates plus average decision time are reported.

## Game definition

The board is a flat list of 9 cells (`"X"`, `"O"` or `" "`). The class provides the
minimal interface the search algorithms need: `possible_moves`, `make_move`,
`switch_player`, `game_over`, `winner`, `score` and `clone`.

```python
import random
import time
```

```python
class TicTacDoh:
    """Tic-Tac-Toe variant used as the search domain for the algorithms below."""

    def __init__(self):
        self.board = [" "] * 9
        self.current_player = "X"

    def display(self):
        for i in range(3):
            print("|".join(self.board[i*3:(i+1)*3]))
            if i < 2:
                print("-"*5)

    def possible_moves(self):
        return [i for i, cell in enumerate(self.board) if cell == " "]

    def make_move(self, move):
        self.board[move] = self.current_player

    def switch_player(self):
        self.current_player = "O" if self.current_player == "X" else "X"

    def game_over(self):
        return self.winner() is not None or " " not in self.board

    def winner(self):
        wins = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8],
            [0, 3, 6], [1, 4, 7], [2, 5, 8],
            [0, 4, 8], [2, 4, 6]
        ]
        for combo in wins:
            a, b, c = combo
            if self.board[a] == self.board[b] == self.board[c] != " ":
                return self.board[a]
        return None

    def clone(self):
        new_game = TicTacDoh()
        new_game.board = self.board[:]
        new_game.current_player = self.current_player
        return new_game

    def score(self):
        winner = self.winner()
        if winner == "X":
            return 1
        elif winner == "O":
            return -1
        else:
            return 0
```

## Search algorithms

**Negamax** is a minimax formulation from the perspective of the player to move:
the value of a node is the negation of the best value the *opponent* can achieve
one ply deeper. With no evaluation function, non-terminal states at the depth
cutoff are scored as 0.

```python
def negamax(game, depth, player):
    """Plain negamax. Returns (value, best_move) from the perspective of `player`."""
    if depth == 0 or game.game_over():
        return player * game.score(), None

    best_value = float('-inf')
    best_move = None
    for move in game.possible_moves():
        new_game = game.clone()
        new_game.make_move(move)
        new_game.switch_player()
        val, _ = negamax(new_game, depth - 1, -player)
        val = -val
        if val > best_value:
            best_value = val
            best_move = move
    return best_value, best_move
```

The alpha-beta version adds a move-pruning window (`alpha`, `beta`): branches that
cannot influence the final decision are cut off early, giving identical results with
significantly fewer node evaluations.

```python
def negamax_ab(game, depth, alpha, beta, player):
    """Negamax with alpha-beta pruning. Returns (value, best_move), equivalent to negamax."""
    if depth == 0 or game.game_over():
        return player * game.score(), None

    best_value = float('-inf')
    best_move = None
    for move in game.possible_moves():
        new_game = game.clone()
        new_game.make_move(move)
        new_game.switch_player()
        val, _ = negamax_ab(new_game, depth - 1, -beta, -alpha, -player)
        val = -val
        if val > best_value:
            best_value = val
            best_move = move
        alpha = max(alpha, val)
        if alpha >= beta:
            break
    return best_value, best_move
```

## Experiment

Each configuration plays `games_count` games with the starting player alternating
every game. In the *probabilistic* variant, each chosen move fails with 20%
probability and the turn passes to the opponent. The RNG is seeded so the
probabilistic results are reproducible.

```python
def run_comparison(games_count=1000):
    """Play self-play games for each algorithm x variant x depth and report results."""
    random.seed(42)
    algorithms = {
        "Negamax": lambda g, d, p: negamax(g, d, p),
        "NegamaxAB": lambda g, d, p: negamax_ab(g, d, float('-inf'), float('inf'), p)
    }
    for algo_name, algo_fn in algorithms.items():
        print(f"\nAlgorytm: {algo_name}")
        for probabilistic in [False, True]:
            print(f"Wariant {'probabilistyczny' if probabilistic else 'deterministyczny'}:")
            for depth in [2, 4]:
                results = {"X": 0, "O": 0, "Draw": 0}
                total_time = 0
                total_moves = 0
                for i in range(games_count):
                    game = TicTacDoh()
                    game.current_player = "X" if i % 2 == 0 else "O"
                    while not game.game_over():
                        start = time.perf_counter()
                        _, move = algo_fn(game, depth, 1 if game.current_player == "X" else -1)
                        total_time += time.perf_counter() - start
                        total_moves += 1

                        if move is not None:
                            if probabilistic and random.random() < 0.2:
                                # Ruch nieudany – przeciwnik przechodzi do ruchu
                                game.switch_player()
                                continue

                            game.make_move(move)

                            if not game.game_over():
                                game.switch_player()
                        else:
                            break
                    winner = game.winner()
                    if winner:
                        results[winner] += 1
                    else:
                        results["Draw"] += 1
                avg_time = total_time / total_moves if total_moves > 0 else 0
                print(f"  Głębokość {depth} → X: {results['X']} | O: {results['O']} | Remisy: {results['Draw']} | Średni czas decyzji: {avg_time:.6f} s")
```

## Results

```python
run_comparison()
```
